## The question

**Decision:** Which visible pages should be reviewed first for a possible CTR or engagement improvement?

**Action:** A content or SEO reviewer could inspect the highest-ranked pages and decide whether to test a title/meta change, improve intent matching or snippet structure, improve on-page engagement, or simply monitor the page.

**Cost of a wrong call:** Ranking an unnecessary page too highly could waste review time. Ranking a genuinely underperforming page too low could cause a useful improvement opportunity to be missed. Therefore, the output should support human review rather than automatically deciding what to change.
