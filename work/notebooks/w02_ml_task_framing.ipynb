## Why ML beats a fixed rule here

A fixed rule could say that every page below a particular CTR threshold should be reviewed. The problem is that CTR depends on several factors, including search position and impressions, so one threshold may treat very different pages the same way.

ML can learn combinations of these signals and produce a ranked list of pages instead of relying on one manually chosen threshold. This makes the output more useful for prioritizing human review. The model would support the decision rather than automatically deciding what changes should be made.
